## 1. 1부터 n까지의 합 (하)

정수 `n`을 받아 1부터 `n`까지의 합을 돌려주는 함수 `total(n)`을 만든다.

````python
def total(n):
    s = 0
    for i in range(1, n + 1):
        s += i
    return s

print(total(10))    # 55
print(total(100))   # 5050
````

## 2. 리스트의 최댓값 (하)

리스트를 받아 가장 큰 값을 돌려주는 함수 `my_max(lst)`를 만든다.

````python
def my_max(lst):
    m = lst[0]
    for x in lst:
        if x > m:
            m = x
    return m

print(my_max([3, 7, 2, 9, 4]))   # 9
print(my_max([-5, -2, -8]))      # -2
````

## 3. 값의 개수 세기 - 기본값 (하)

리스트 `lst`에서 `target`과 같은 원소의 개수를 돌려주는 함수 `count_value(lst, target=0)`을 만든다.
`target`을 주지 않으면 0의 개수를 센다.

````python
def count_value(lst, target=0):
    cnt = 0
    for x in lst:
        if x == target:
            cnt += 1
    return cnt

print(count_value([0, 1, 0, 2, 0]))      # 3
print(count_value([1, 2, 1, 3], 1))      # 2
````

## 4. 직각삼각형 출력 - 기본값 2개 (하)

`n`줄짜리 직각삼각형을 문자 `ch`로 출력하는 함수 `triangle(n, ch='*')`를 만든다.
`i`번째 줄에는 `ch`가 `i`개 있다.

````python
def triangle(n, ch='*'):
    for i in range(1, n + 1):
        line = ''
        for j in range(i):
            line += ch
        print(line)

triangle(3)
print()
triangle(4, '#')
````

## 5. 소수 판별 (중)

정수 `n`이 소수이면 `True`, 아니면 `False`를 돌려주는 함수 `is_prime(n)`을 만든다.
소수는 1과 자기 자신만을 약수로 가지는 2 이상의 정수다.

````python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

print(is_prime(7))    # True
print(is_prime(12))   # False
print(is_prime(1))    # False
````

## 6. n 이하의 소수 목록 - 함수 안에서 함수 호출 (중)

5번의 `is_prime`을 사용해서, 2부터 `n`까지의 소수를 리스트로 돌려주는 함수 `primes_upto(n)`을 만든다.

````python
def primes_upto(n):
    result = []
    for i in range(2, n + 1):
        if is_prime(i):
            result.append(i)
    return result

print(primes_upto(20))   # [2, 3, 5, 7, 11, 13, 17, 19]
````

## 7. 지역변수 - 실행 결과 예측 (하)

**(2)**

````python
def g(a):
    a = a + 1
    return a

a = 3
b = g(a)
print(a, b)
````

결과: `3 4`

매개변수 `a`도 지역변수다. 함수 안에서 `a`를 바꿔도 바깥의 `a`는 그대로다.

**(3)**

결과: `NameError`

`t`는 `f2` 안에서 만든 지역변수이므로 함수 밖에서는 없다.

## 8. 호출 횟수 세기 - global (중)

함수 `hello()`가 호출될 때마다 '안녕'을 출력하고, 전역변수 `call_count`를 1씩 늘린다.

먼저 `global` 없이 실행해 본다.

````python
call_count = 0

def hello():
    call_count += 1
    print('안녕')

hello()
````

결과: `UnboundLocalError`

함수 안에 `call_count += 1`처럼 이름에 대입하는 문장이 있으면, `call_count`는 그 함수 전체에서 지역변수로 처리된다.
지역변수 `call_count`에는 아직 값이 없으므로 1을 더할 수 없다.

함수 안에서 `global call_count`라고 쓰면 전역변수 `call_count`에 대입한다.

````python
call_count = 0

def hello():
    global call_count
    call_count += 1
    print('안녕')

hello()
hello()
hello()
print(call_count)   # 3
````

## 9. 합이 target인 쌍의 개수 (중)

리스트 `lst`에서 서로 다른 두 위치 `i < j`를 골랐을 때 `lst[i] + lst[j] == target`인 쌍의 개수를 돌려주는 함수 `count_pairs(lst, target)`을 만든다.

````python
def count_pairs(lst, target):
    cnt = 0
    for i in range(len(lst)):
        for j in range(i + 1, len(lst)):
            if lst[i] + lst[j] == target:
                cnt += 1
    return cnt

print(count_pairs([1, 2, 3, 4, 5], 6))   # 2  (1+5, 2+4)
print(count_pairs([3, 3, 3], 6))         # 3
````

`cnt`는 함수 안에서 만든 지역변수이므로 `global`이 필요 없다.

## 10. 버블 정렬과 교환 횟수 - global (중)

리스트를 오름차순으로 정렬하는 함수 `bubble_sort(lst)`를 만든다.

- 버블 정렬: 앞에서부터 이웃한 두 값을 비교해서 앞의 값이 더 크면 서로 바꾼다. 이것을 리스트 길이 - 1 번 반복한다.
- 함수는 값을 돌려주지 않고, 받은 리스트 자체를 정렬한다.
- 두 값을 바꿀 때마다 전역변수 `swap_count`를 1씩 늘린다.

````python
swap_count = 0

def bubble_sort(lst):
    global swap_count
    n = len(lst)
    for i in range(n - 1):
        for j in range(n - 1 - i):
            if lst[j] > lst[j + 1]:
                lst[j], lst[j + 1] = lst[j + 1], lst[j]
                swap_count += 1

data = [5, 3, 4, 1, 2]
bubble_sort(data)
print(data)         # [1, 2, 3, 4, 5]
print(swap_count)   # 8
````

- `swap_count += 1`은 이름 `swap_count`에 대입하므로 `global`이 필요하다.
- `lst[j], lst[j + 1] = ...`은 이름 `lst`에 대입하는 것이 아니라 `lst`가 가리키는 리스트의 내용을 바꾸는 것이므로 `global`이 필요 없다. 그래서 함수 밖의 `data`가 정렬된다.
