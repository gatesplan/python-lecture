# 10. 알고리즘 - 퀵 정렬

퀵 정렬은 구간 하나를 다음 세 단계로 처리하고 이를 재귀로 반복한다.

1. 구간에서 **피벗**(기준값) 하나를 정한다.
2. 피벗보다 작은 값을 왼쪽, 큰 값을 오른쪽으로 몰고 피벗을 그 경계에 놓는다. 이 과정을 **분할**(partition)이라 한다.
3. 피벗 왼쪽 구간과 오른쪽 구간에 같은 일을 반복한다.

분할이 끝난 피벗은 최종 자리에 있다. 그 뒤로 다시 움직이지 않는다.

분할을 구현하는 방법은 여러 가지다. 이 자료는 셋을 다룬다.

| 절 | 분할 방식 | 커서 | 원본 리스트 |
| --- | --- | --- | --- |
| 1 | 양쪽 커서 | `i`는 오른쪽으로, `j`는 왼쪽으로 | 직접 수정 |
| 2 | 한쪽 커서 | `i`, `j` 모두 오른쪽으로 | 직접 수정 |
| 3 | 리스트 컴프리헨션 | 없음 | 수정하지 않음 |

각 원소가 단계마다 어떻게 움직이는지는 같은 폴더의 `10. 알고리즘 - 퀵 정렬.html`에서 애니메이션으로 볼 수 있다.

## 1. 양쪽 커서

구간 `A[lo..hi]`를 분할한다.

- 피벗 `p = A[hi]`. 구간의 오른쪽 끝 값이다.
- 커서 `i`는 `lo`에서 오른쪽으로, 커서 `j`는 `hi-1`에서 왼쪽으로 이동한다.
- `i`는 `A[i] < p`인 동안 민다. **피벗보다 크거나 같은 값에서 멈춘다.**
- `j`는 `A[j] > p`인 동안 민다. **피벗보다 작거나 같은 값에서 멈춘다.**
- 둘 다 멈췄을 때 `i < j`이면 `A[i]`와 `A[j]`를 맞바꾸고 두 커서를 한 칸씩 민다.
- `i >= j`이면 반복을 끝내고 `A[i]`와 피벗 `A[hi]`를 맞바꾼다. 피벗의 최종 자리는 `i`다.

````python
def part(A, lo, hi):
    p = A[hi]                        # 피벗 = 구간의 오른쪽 끝 값
    i = lo                           # 왼쪽 커서
    j = hi - 1                       # 오른쪽 커서

    while True:
        while i <= j and A[i] < p:   # 피벗 이상인 값에서 멈춘다
            i += 1
        while i <= j and A[j] > p:   # 피벗 이하인 값에서 멈춘다
            j -= 1

        if i >= j:                   # 커서가 만나거나 엇갈리면 분할 끝
            break

        A[i], A[j] = A[j], A[i]      # 반대편에 있어야 할 두 값을 맞바꾼다
        i += 1
        j -= 1

    A[i], A[hi] = A[hi], A[i]        # 피벗을 경계 자리로
    return i                         # 피벗이 확정된 자리


def quick(A, lo, hi):
    if lo >= hi:                     # 길이 1 이하인 구간은 할 일이 없다
        return

    m = part(A, lo, hi)              # 피벗이 m번 자리에 확정된다
    quick(A, lo, m - 1)              # 피벗 왼쪽 구간
    quick(A, m + 1, hi)              # 피벗 오른쪽 구간


A = [5, 3, 8, 4, 2, 7, 1, 6]
quick(A, 0, len(A) - 1)
print(A)
````

```
[1, 2, 3, 4, 5, 6, 7, 8]
```

## 코드 읽기

`part`가 분할 한 번을 맡고, `quick`이 재귀를 맡는다.

- `p = A[hi]` - 피벗을 값으로 미리 꺼내 둔다. 분할하는 동안 `A[hi]` 자리는 건드리지 않는다.
- `j = hi - 1` - 피벗이 있는 `hi`는 훑을 대상이 아니므로 한 칸 안에서 시작한다.
- 안쪽 `while` 두 개 - 커서를 민다. 조건에 걸리는 값을 만나면 멈춘다.
  두 커서가 모두 멈췄다는 것은 **왼쪽에 큰 값이, 오른쪽에 작은 값이 있다**는 뜻이다. 그래서 맞바꾼다.
- `i <= j` - 두 커서가 서로를 넘어가지 못하게 막는다. 이 조건이 없으면 커서가 구간 밖으로 나가 다른 구간의 값을 읽는다.
- `if i >= j: break` - 커서가 만나거나 엇갈리면 더 바꿀 것이 없다.
- `A[i], A[hi] = A[hi], A[i]` - 반복이 끝난 시점에 `i`는 피벗 이상인 값들 중 가장 왼쪽 자리다.
  그 자리와 피벗을 바꾸면 피벗의 왼쪽은 전부 피벗 이하, 오른쪽은 전부 피벗 이상이 된다.
- `return i` - `quick`은 이 값을 받아 왼쪽과 오른쪽 구간을 나눈다. `m`번 자리는 재귀 범위에서 빠진다.
- `quick`의 `if lo >= hi: return` - 이것이 재귀의 기저 조건이다.

`quick(A, 0, len(A)-1)`처럼 구간을 인자로 넘긴다. 반환값은 없고 `A`가 직접 바뀐다.

## 2. 한쪽 커서

같은 분할을 커서 하나로 한다. 피벗은 그대로 `A[hi]`를 쓴다.

- `i`는 "피벗보다 작은 값이 채워질 다음 자리"다. `lo`에서 시작한다.
- `j`를 `lo`부터 `hi-1`까지 한 방향으로 끝까지 훑는다. 멈추지 않는다.
- `A[j] < p`이면 `A[i]`와 맞바꾸고 `i`를 한 칸 민다.
- 훑기가 끝나면 `A[i]`와 피벗을 맞바꾼다. 피벗의 자리는 `i`다.

````python
def part2(A, lo, hi):
    p = A[hi]                        # 피벗 = 구간의 오른쪽 끝 값
    i = lo                           # 작은 값이 채워질 다음 자리

    for j in range(lo, hi):          # 구간을 왼쪽에서 오른쪽으로 끝까지 훑는다
        if A[j] < p:
            A[i], A[j] = A[j], A[i]  # 작은 값을 i 자리로 보낸다
            i += 1

    A[i], A[hi] = A[hi], A[i]        # 피벗을 경계 자리로
    return i


def quick2(A, lo, hi):
    if lo >= hi:
        return

    m = part2(A, lo, hi)
    quick2(A, lo, m - 1)
    quick2(A, m + 1, hi)


A = [5, 3, 8, 4, 2, 7, 1, 6]
quick2(A, 0, len(A) - 1)
print(A)
````

```
[1, 2, 3, 4, 5, 6, 7, 8]
```

## 코드 읽기

`quick2`는 `quick`과 완전히 같다. 달라진 것은 `part2` 하나뿐이다.

`for`문이 도는 동안 구간은 항상 다음 상태를 유지한다.

```
lo    ..  i-1  : 피벗보다 작은 값
i     ..  j-1  : 피벗보다 크거나 같은 값
j     ..  hi-1 : 아직 안 본 값
hi             : 피벗
```

- `A[j] < p`이면 `A[j]`는 왼쪽 덩어리로 가야 한다. 그 자리가 `i`다. 맞바꾸고 `i`를 민다.
- `A[j] >= p`이면 이미 오른쪽 덩어리에 속하므로 아무것도 하지 않는다.
- `i == j`일 때의 교환은 제자리 교환이라 아무 일도 일어나지 않는다.

## 양쪽 커서와 비교

- 코드가 짧고 커서가 하나뿐이라 읽기 쉽다.
- 교환 횟수는 더 많다. 피벗보다 작은 값을 만날 때마다 교환하므로 **이미 제자리에 있는 값도 교환한다.**
  양쪽 커서 방식은 서로 반대편에 있는 값끼리만 교환한다.
- 피벗이 놓이는 자리는 값이 모두 다르면 두 방식이 같다.
  피벗과 같은 값이 섞이면 달라진다. 양쪽 커서 방식은 피벗과 같은 값에서도 커서가 멈춰 교환하기 때문이다.
- 정렬 결과는 어느 쪽이든 같다.

## 3. 리스트 컴프리헨션

같은 아이디어를 제자리 교환 없이 쓴다. 피벗보다 작은 것, 같은 것, 큰 것을 새 리스트 셋으로 나눈다.
분할과 재귀가 한 함수 안에 들어가므로 `part`가 따로 없다.

````python
def quick3(A):
    if len(A) <= 1:                   # 길이 1 이하는 이미 정렬된 상태
        return A

    p = A[len(A)//2]                  # 피벗 = 가운데 자리 값
    L = [x for x in A if x < p]       # 피벗보다 작은 값
    M = [x for x in A if x == p]      # 피벗과 같은 값
    R = [x for x in A if x > p]       # 피벗보다 큰 값

    return quick3(L) + M + quick3(R)  # 정렬한 L, 피벗, 정렬한 R 을 이어 붙인다


A = [5, 3, 8, 4, 2, 7, 1, 6]
print(quick3(A))
print(A)
````

```
[1, 2, 3, 4, 5, 6, 7, 8]
[5, 3, 8, 4, 2, 7, 1, 6]
```

## 코드 읽기

- `M`을 따로 두는 이유는 피벗과 같은 값이 여러 개일 수 있기 때문이다.
  이 값들은 이미 제자리가 정해졌으므로 재귀에서 뺀다. `M`이 없으면 같은 값이 많을 때 리스트 크기가 줄지 않아 무한 재귀가 된다.
- `quick3(L) + M + quick3(R)`에서 파이썬은 왼쪽부터 계산한다. `L` 쪽이 끝까지 처리된 뒤에 `R` 쪽이 시작된다.

## 앞의 두 방식과 비교

- **원본을 수정하지 않는다.** 두 번째 `print(A)`가 원래 리스트를 그대로 출력한다.
  결과를 쓰려면 `A = quick3(A)`라고 받아야 한다. 이 실수가 흔하다.
- 호출 형태가 다르다. `quick3(A)` 하나면 되고 구간을 넘기지 않는다.
- 매 단계 새 리스트를 만들므로 추가 메모리가 필요하다. 제자리 정렬이 아니다.
- 교환이 없으므로 값이 리스트 안에서 이동하는 모습을 볼 수 없다.

## 4. 피벗 선택

피벗을 어디서 고르든 결과는 정렬된 리스트다. 달라지는 것은 재귀 깊이다.

- **좋은 피벗** - 왼쪽과 오른쪽이 비슷한 크기로 갈린다. 깊이가 log n이 된다.
- **나쁜 피벗** - 한쪽이 비게 갈린다. 깊이가 n이 된다. 구간의 최솟값이나 최댓값이 피벗일 때 그렇다.

1, 2절처럼 오른쪽 끝을 피벗으로 쓰면 **이미 정렬된 리스트**에서 매번 최댓값이 피벗이 되어 깊이가 n이 된다.
3절처럼 가운데를 쓰면 이 경우는 피하지만, 가운데 값이 최솟값이나 최댓값이 되는 다른 입력에서 같은 일이 생긴다.
**어떤 고정된 규칙이든 그 규칙을 망가뜨리는 입력이 존재한다.**

실무 구현은 피벗을 무작위로 고르거나, 첫/가운데/마지막 세 값의 중간값을 쓴다.
가운데 값을 쓰고 싶은데 `part`는 `A[hi]`를 피벗으로 삼는다면, 분할하기 전에 가운데 값과 `A[hi]`를 맞바꾸면 된다.
`part` 자체는 고치지 않아도 된다.

파이썬의 재귀 깊이 한계는 기본 1000 근처다. 최악의 경우 깊이가 n까지 가므로
원소가 1000개를 넘는 리스트에서는 `RecursionError`가 날 수 있다.

## 5. 복잡도

| 경우 | 시간복잡도 | 조건 |
| --- | --- | --- |
| 최선 | O(n log n) | 매번 절반에 가깝게 갈릴 때 |
| 평균 | O(n log n) | 임의의 입력 |
| 최악 | O(n^2) | 매번 한쪽이 비게 갈릴 때 |

한 번 분할할 때 그 구간의 원소를 전부 한 번씩 본다. 같은 깊이에 있는 분할들의 작업량을 합치면 n에 비례한다.
따라서 전체 작업량은 **n 곱하기 깊이**다.

- 깊이가 log n이면 O(n log n)
- 깊이가 n이면 O(n^2)

공간복잡도는 1, 2절이 재귀 호출 스택만 쓰므로 O(log n) ~ O(n)이다.
3절은 매 단계 새 리스트를 만들므로 O(n log n)까지 쓴다.

## 안정성

값이 같은 원소의 원래 순서가 정렬 후에도 유지되면 안정 정렬이다.

- 1, 2절 - 멀리 떨어진 두 자리를 맞바꾸므로 사이에 있던 같은 값의 순서가 깨진다. **불안정.**
- 3절 - 리스트를 앞에서부터 훑어 담으므로 순서가 유지된다. **안정.**

같은 퀵 정렬이라도 구현에 따라 안정성이 달라진다.

## 6. 세 방식 비교

| | 양쪽 커서 | 한쪽 커서 | 리스트 컴프리헨션 |
| --- | --- | --- | --- |
| 피벗 | 오른쪽 끝 | 오른쪽 끝 | 가운데 |
| 커서 | `i` 오른쪽으로, `j` 왼쪽으로 | `i`, `j` 모두 오른쪽으로 | 없음 |
| 원본 | 직접 수정 | 직접 수정 | 수정 안 함 |
| 분할 함수 반환값 | 피벗 자리 | 피벗 자리 | 없음 (정렬된 새 리스트를 반환) |
| 추가 메모리 | 거의 없음 | 거의 없음 | O(n log n) |
| 교환 횟수 | 적음 | 많음 | 교환 없음 |
| 피벗과 같은 값 | 커서가 멈춰 교환한다 | 피벗 오른쪽으로 몰린다 | `M`으로 분리해 재귀에서 제외 |
| 안정 | 아니오 | 아니오 | 예 |
| 호출 형태 | `quick(A, 0, len(A)-1)` | `quick2(A, 0, len(A)-1)` | `A = quick3(A)` |

세 방식 모두 같은 결과를 낸다.
