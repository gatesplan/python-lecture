# 8. 복습 - 스택과 큐

1~3강 내용을 난이도 순으로 다시 정리한 노트북. 새 내용은 없다.

| 단계 | 주제 |
|------|------|
| 1 | 개념과 기본 동작 |
| 2 | 세 가지 구현 방식 |
| 3 | 핵심 성질 |
| 4 | 기본 응용 |
| 5 | 도전 |

2단계에서 클래스 구현을 한 번 보지만, **3~5단계 코드는 전부 리스트를 스택/큐로 직접 쓴다.**
알고리즘을 짤 때는 자료구조를 따로 만들지 않고 `append` / `pop` / `pop(0)` 으로 바로 쓰는 것이 보통이다.

위에서부터 순서대로 실행하며 읽는다.

# 1단계. 개념과 기본 동작

## 1-1. 정의

**스택**: 맨 뒤에 넣고 맨 뒤에서 뺀다. LIFO (Last In First Out). 접시 더미.

```
push(10) -> [10]
push(20) -> [10, 20]
pop()    -> 20 반환, [10]
```

**큐**: 맨 뒤에 넣고 맨 앞에서 뺀다. FIFO (First In First Out). 줄 서기.

```
enqueue(10) -> [10]
enqueue(20) -> [10, 20]
dequeue()   -> 10 반환, [20]
```

## 1-2. 비교

| | 스택 | 큐 |
|---|---|---|
| 규칙 | LIFO | FIFO |
| 넣기 | push | enqueue |
| 빼기 | pop | dequeue |
| 넣는 위치 | 맨 뒤 | 맨 뒤 |
| **빼는 위치** | **맨 뒤** | **맨 앞** |
| 확인만 | peek | front |

넣는 위치는 같다. 다른 것은 빼는 위치뿐이다.

## 1-3. 리스트로 스택 쓰기

````python
s = []
s.append(10)          # push
s.append(20)
s.append(30)

print(s.pop())        # 30     맨 뒤를 빼서 반환
print(s)              # [10, 20]
print(s[-1])          # 20     peek
print(len(s) == 0)    # False  is_empty
print(len(s))         # 2      size
````

`pop()` 은 인자 없이 부르면 맨 뒤를 뺀다. `s[-1]` 이 peek.

## 1-4. 리스트로 큐 쓰기

````python
q = []
q.append("가")        # enqueue
q.append("나")
q.append("다")

print(q.pop(0))       # 가     맨 앞을 빼서 반환
print(q)              # ['나', '다']
print(q[0])           # 나     front
````

`pop()` 의 인자 `0` 하나가 스택과 큐를 가른다.

```
s.pop()    -> 맨 뒤 (스택)
q.pop(0)   -> 맨 앞 (큐)
```

리스트 하나로 스택도 큐도 된다. 이름을 붙인 클래스가 따로 필요하지 않다.

## 1-5. 넣는 코드는 같고 나오는 값은 반대

````python
s = []
s.append(1)
s.append(2)
print("스택:", s.pop())   # 2

q = []
q.append(1)
q.append(2)
print("큐  :", q.pop(0))  # 1
````

# 2단계. 세 가지 구현 방식

| 방식 | 자료와 동작 | 명칭 |
|------|-----------|------|
| A. 전역 리스트 + 함수 | 분리 (전역 공유) | 절차적 |
| B. 리스트 전달 + 함수 | 분리 (인자 전달) | 절차적 |
| C. 클래스 | 결합 (캡슐화) | 객체지향 |

셋 다 하는 일은 같다. 껍데기만 다르다. 시험 코드는 어느 방식으로든 나오므로 읽을 줄 알아야 한다.

## 2-1. 방식 A - 전역 리스트

````python
stack = []          # 전역 스택

def push(x):
    stack.append(x)

def pop():
    return stack.pop()

def peek():
    return stack[-1]


push(1)
push(2)
push(3)
print(pop())     # 3
push(4)
print(peek())    # 4
print(stack)     # [1, 2, 4]
````

한계: 전역 `stack` 하나에 묶여 스택을 두 개 쓸 수 없다. 어디서든 건드릴 수 있어 오염되기 쉽다.
교과서와 학교 시험이 자주 쓴다.

## 2-2. 방식 B - 리스트 전달

````python
def enqueue(queue, x):
    queue.append(x)

def dequeue(queue):
    return queue.pop(0)


a = []
b = []
enqueue(a, 1)
enqueue(a, 2)
enqueue(b, 99)
print(dequeue(a))   # 1
print(a)            # [2]
print(b)            # [99]
````

`a` 와 `b` 가 독립적으로 동작한다.
리스트를 넘기면 같은 리스트를 가리키므로(참조), 함수 안의 `append` 가 바깥에도 반영된다.

## 2-3. 방식 C - 클래스

자료(`self.items`)와 동작(메서드)을 묶는다. `self` 는 객체 자신, `__init__` 은 생성 시 실행된다.

````python
class Stack:
    def __init__(self):
        self.items = []
    def push(self, x):
        self.items.append(x)
    def pop(self):
        return self.items.pop()
    def peek(self):
        return self.items[-1]
    def is_empty(self):
        return len(self.items) == 0
    def size(self):
        return len(self.items)


s = Stack()
for x in [5, 10, 15]:
    s.push(x)

print(s.pop() + s.pop())   # 25   15 + 10
print(s.size())            # 1

t = Stack()                # 독립적인 또 하나의 스택
t.push(100)
print(s.items, t.items)    # [5] [100]
````

## 2-4. 정리

| 방식 | 여러 개 사용 | 장점 | 단점 |
|------|------------|------|------|
| A. 전역 | 불가능 | 짧다 | 하나에 묶임, 오염 가능 |
| B. 전달 | 가능 | 전역 오염 없음 | 매번 전달 |
| C. 클래스 | 가능 | 자료+동작 결합 | 문법이 길다 |

낯선 코드를 만나면 "결국 `append` 와 `pop` 이다"까지 벗겨내면 된다.

**아래 3~5단계는 리스트를 그대로 스택/큐로 쓴다.** 알고리즘 코드에서는 이 형태가 기본이다.

# 3단계. 핵심 성질

4~5단계 응용이 전부 여기서 나온다.

## 3-1. 스택은 순서를 뒤집는다

````python
s = []
for ch in "STACK":
    s.append(ch)

out = ""
while len(s) > 0:
    out += s.pop()
print(out)             # KCATS
````

전부 넣었다 전부 빼면 역순이 된다. 문자열 뒤집기와 진법 변환이 이 성질로 풀린다.

## 3-2. 큐 회전 - 맨 앞을 맨 뒤로

````python
q = [1, 2, 3, 4]
q.append(q.pop(0))     # 1이 맨 뒤로
print(q)               # [2, 3, 4, 1]
q.append(q.pop(0))     # 2가 맨 뒤로
print(q)               # [3, 4, 1, 2]
````

안쪽 `pop(0)` 이 맨 앞을 빼고, 바깥 `append` 가 맨 뒤에 붙인다. 카드 버리기에서 쓴다.

## 3-3. 문자열을 앞에 붙이나 뒤에 붙이나

````python
n = 13
front = ''
while n > 0:
    front = str(n % 2) + front   # 앞에 붙임
    n //= 2

n = 13
back = ''
while n > 0:
    back = back + str(n % 2)     # 뒤에 붙임
    n //= 2

print(front)   # 1101  맞음
print(back)    # 1011  뒤집힘
````

나머지는 낮은 자리부터 나온다. 앞에 붙이면 높은 자리가 왼쪽으로 밀려간다.

**앞에 붙이기 = 스택에 쌓았다 꺼내기.**

## 3-4. `while n > 0` 은 `n == 0` 을 놓친다

````python
for n in [13, 0]:
    s = ''
    while n > 0:
        s = str(n % 2) + s
        n //= 2
    print(repr(s))     # '1101' / ''
````

`n = 0` 이면 조건이 처음부터 거짓이라 루프에 한 번도 들어가지 않는다. `'0'` 을 따로 처리해야 한다.

## 3-5. 단락 평가(short-circuit)

`or` 는 왼쪽이 참이면 오른쪽을 실행하지 않는다.

````python
s = []
print(len(s) == 0 or s.pop() != '(')   # True, s.pop()은 실행 안 됨
````

````python
# 순서를 바꾸면 터진다 (일부러 에러를 내는 셀)
s = []
print(s.pop() != '(' or len(s) == 0)   # IndexError
````

괄호 검사에서 `len(s) == 0 or s.pop() != ...` 처럼 **빈 검사를 먼저** 두는 이유다.

# 4단계. 기본 응용

여기부터 리스트를 스택/큐로 직접 쓴다.

## 4-1. 문자열 뒤집기

3-1을 함수로 만든 것.

````python
def reverse_string(text):
    s = []
    for ch in text:
        s.append(ch)

    result = ""
    while len(s) > 0:
        result += s.pop()
    return result


print(reverse_string("hello"))     # olleh
print(reverse_string("스택연습"))   # 습연택스
````

## 4-2. 괄호 검사 - 한 종류 `()`

여는 괄호는 쌓고, 닫는 괄호를 만나면 하나 뺀다.

````python
def is_balanced(text):
    s = []
    for ch in text:
        if ch == "(":
            s.append(ch)
        elif ch == ")":
            if len(s) == 0:
                return False      # 닫는 괄호가 먼저 나옴
            s.pop()
    return len(s) == 0            # 안 닫힌 괄호가 남으면 False


print(is_balanced("(())"))   # True
print(is_balanced("(()"))    # False
print(is_balanced(")("))     # False
````

실패 조건이 두 가지다. 하나만 검사하면 `")("` 또는 `"(()"` 중 하나를 놓친다.

## 4-3. 괄호 검사 - 세 종류 `()` `[]` `{}`

닫는 괄호는 스택 맨 위와 종류까지 맞아야 한다.

````python
def is_balanced(text):
    pairs = {')': '(', ']': '[', '}': '{'}
    opens = "([{"
    s = []
    for ch in text:
        if ch in opens:
            s.append(ch)
        elif ch in pairs:
            if len(s) == 0 or s.pop() != pairs[ch]:
                return False
    return len(s) == 0


print(is_balanced("([{}])"))   # True
print(is_balanced("([)]"))     # False
print(is_balanced("(()"))      # False
````

`"([)]"` 가 False인 과정:

| 글자 | 동작 | 스택 |
|------|------|------|
| `(` | push | `(` |
| `[` | push | `( [` |
| `)` | pop한 `[` 와 짝 `(` 가 다름 -> False | |

가장 최근에 연 괄호부터 닫아야 한다. "가장 최근" = 스택의 맨 위.

## 4-4. 진법 변환 - 10진수 -> n진수 문자열

나머지를 스택에 쌓았다 꺼내면 자리 순서가 맞는다. 10 이상은 문자로: 10->A ... 15->F

```
13을 2진법으로
13 -> 6 ... 1     [1]
 6 -> 3 ... 0     [1, 0]
 3 -> 1 ... 1     [1, 0, 1]
 1 -> 0 ... 1     [1, 0, 1, 1]   몫이 0이라 중단
꺼내며 이어붙임 -> 1101
```

````python
def to_base(n, base):
    digits = "0123456789ABCDEF"
    if n == 0:
        return "0"                    # 3-4의 경계값

    s = []
    while n > 0:
        s.append(digits[n % base])    # 낮은 자리부터 쌓임
        n //= base

    result = ""
    while len(s) > 0:
        result += s.pop()             # 높은 자리부터 나옴
    return result


print(to_base(13, 2))     # 1101
print(to_base(255, 16))   # FF
print(to_base(0, 2))      # 0
````

3-3에 따라 스택 없이 `result = digits[n % base] + result` 로도 같은 결과가 나온다.

## 4-5. 진법 역변환 - n진수 문자열 -> 10진수

높은 자리부터 순서대로 써야 하므로 큐(FIFO)다.

```
'1021' (3진수)
t = 0
1 -> t = 0*3 + 1 = 1
0 -> t = 1*3 + 0 = 3
2 -> t = 3*3 + 2 = 11
1 -> t = 11*3 + 1 = 34
```

한 자리 꺼낼 때마다 진법을 곱해 자리를 올리고 새 자릿수를 더한다.

````python
def from_base(text, base):
    digits = "0123456789ABCDEF"
    q = []
    for ch in text:
        q.append(digits.index(ch))   # 문자 -> 숫자

    t = 0
    while len(q) > 0:
        t = base * t + q.pop(0)
    return t


print(from_base("1021", 3))   # 34
print(from_base("1101", 2))   # 13
print(from_base("FF", 16))    # 255
````

## 4-6. 카드 버리기 (큐)

1~N 카드가 순서대로 있다(1이 맨 위). 한 장 남을 때까지 반복한다.

1. 맨 위 카드를 버린다.
2. 그다음 카드를 맨 아래로 옮긴다.

N = 6 추적:

| 상태 | 동작 |
|------|------|
| [1,2,3,4,5,6] | 1 버림, 2를 아래로 |
| [3,4,5,6,2] | 3 버림, 4를 아래로 |
| [5,6,2,4] | 5 버림, 6을 아래로 |
| [2,4,6] | 2 버림, 4를 아래로 |
| [6,4] | 6 버림, 4를 아래로 |
| [4] | 답 4 |

````python
def last_card(n):
    q = []
    for i in range(1, n + 1):
        q.append(i)

    while len(q) > 1:
        q.pop(0)               # 맨 위 버리기
        q.append(q.pop(0))     # 다음 카드를 맨 아래로 (3-2의 회전)
    return q[0]


print(last_card(2))   # 2
print(last_card(4))   # 4
print(last_card(6))   # 4
````

반복 조건은 `len(q) > 1`. `> 0` 이면 마지막 한 장까지 버리고 빈 리스트에서 `q[0]` 을 읽어 터진다.

# 5단계. 도전

## 5-1. 중위 -> 후위 변환 (Shunting-yard)

`3 + 4 * 2` (중위) -> `3 4 2 * +` (후위). 계산기가 우선순위를 처리하는 방법. 연산자 스택을 쓴다.
입력은 공백으로 구분된 토큰: `"( 3 + 4 ) * 2"`

규칙:
- 숫자 -> 바로 출력
- 연산자 -> 스택 위 연산자의 우선순위가 크거나 같으면 꺼내 출력, 그다음 자신을 push
- `(` -> push
- `)` -> `(` 가 나올 때까지 꺼내 출력하고 `(` 는 버림
- 끝나면 스택에 남은 것을 모두 꺼내 출력

`3 + 2 * 3` 추적:

| 토큰 | 동작 | 스택 | 출력 |
|------|------|------|------|
| 3 | 출력 |  | 3 |
| + | push | + | 3 |
| 2 | 출력 | + | 3 2 |
| * | `+` 보다 우선순위 높음, push | + * | 3 2 |
| 3 | 출력 | + * | 3 2 3 |
| 끝 | 스택 비우기 |  | 3 2 3 * + |

`( 3 + 2 ) * 3` 추적:

| 토큰 | 동작 | 스택 | 출력 |
|------|------|------|------|
| ( | push | ( |  |
| 3 | 출력 | ( | 3 |
| + | top이 `(` 라 안 꺼냄, push | ( + | 3 |
| 2 | 출력 | ( + | 3 2 |
| ) | `(` 까지 꺼냄, `(` 버림 |  | 3 2 + |
| * | push | * | 3 2 + |
| 3 | 출력 | * | 3 2 + 3 |
| 끝 | 스택 비우기 |  | 3 2 + 3 * |

````python
def to_postfix(expr):
    prec = {'+': 1, '-': 1, '*': 2, '/': 2}
    s = []          # 연산자 스택
    output = []

    for token in expr.split():
        if token in prec:                       # 연산자
            while (len(s) > 0 and
                   s[-1] != '(' and
                   prec[s[-1]] >= prec[token]):
                output.append(s.pop())
            s.append(token)
        elif token == '(':
            s.append(token)
        elif token == ')':
            while s[-1] != '(':
                output.append(s.pop())
            s.pop()                             # '(' 버리기
        else:                                   # 숫자
            output.append(token)

    while len(s) > 0:
        output.append(s.pop())
    return " ".join(output)


print(to_postfix("3 + 2 * 3"))        # 3 2 3 * +
print(to_postfix("( 3 + 2 ) * 3"))    # 3 2 + 3 *
````

### `>=` 와 `>` 의 차이 (결합성)

같은 우선순위끼리 만났을 때 어떻게 하느냐가 갈린다.

````python
def to_postfix2(expr, strict):
    prec = {'+': 1, '-': 1, '*': 2, '/': 2}
    s = []
    out = []
    for t in expr.split():
        if t in prec:
            while len(s) > 0 and (prec[s[-1]] > prec[t] if strict
                                  else prec[s[-1]] >= prec[t]):
                out.append(s.pop())
            s.append(t)
        else:
            out.append(t)
    while len(s) > 0:
        out.append(s.pop())
    return " ".join(out)


print(to_postfix2("5 - 3 - 1", strict=False))   # >= : 5 3 - 1 -
print(to_postfix2("5 - 3 - 1", strict=True))    # >  : 5 3 1 - -
````

- `>=`: `5 3 - 1 -` -> `(5 - 3) - 1 = 1`  (맞음)
- `>` : `5 3 1 - -` -> `5 - (3 - 1) = 3`  (틀림)

같은 우선순위는 왼쪽 것을 먼저 꺼내야(`>=`) 왼쪽부터 계산된다.
`+`, `*` 만 쓰면 결과가 같아 버그가 안 보인다.

## 5-2. 후위식 계산

피연산자 스택을 쓴다.
- 숫자 -> push
- 연산자 -> 두 개 꺼내 계산 후 결과를 push
- 먼저 꺼낸 것이 오른쪽(b), 나중 꺼낸 것이 왼쪽(a). 계산은 `a (연산) b`

````python
def eval_postfix(expr):
    s = []          # 피연산자 스택
    for token in expr.split():
        if token in ("+", "-", "*", "/"):
            b = s.pop()      # 오른쪽 (먼저 꺼냄)
            a = s.pop()      # 왼쪽 (나중 꺼냄)
            if token == "+":
                s.append(a + b)
            elif token == "-":
                s.append(a - b)
            elif token == "*":
                s.append(a * b)
            else:
                s.append(a / b)
        else:
            s.append(int(token))
    return s.pop()


print(eval_postfix("3 4 2 * +"))   # 11
print(eval_postfix("3 4 + 2 *"))   # 14
print(eval_postfix("10 3 -"))      # 7
````

### 두 pop의 순서를 바꾸면

````python
s = [10, 3]
b = s.pop()      # 3
a = s.pop()      # 10
print(a - b)     # 7    맞음

s = [10, 3]
a = s.pop()      # 3    순서를 바꿈
b = s.pop()      # 10
print(a - b)     # -7   틀림
````

`+`, `*` 는 교환법칙이 성립해 순서를 바꿔도 같다. `-`, `/` 에서만 버그가 드러난다.

## 5-3. 둘을 이으면 계산기

````python
for expr in ["( 3 + 4 ) * 2", "3 + 4 * 2", "5 - 3 - 1"]:
    post = to_postfix(expr)
    print(expr, "->", post, "=", eval_postfix(post))
````

## 5-4. 리스트 큐의 `pop(0)` 은 느리다

`pop(0)` 은 맨 앞을 빼고 나머지 전부를 한 칸씩 앞으로 당긴다. 데이터가 n개면 매번 n번 옮긴다.
스택의 `pop()` 은 맨 뒤라서 당길 필요가 없다.

해결: 지우지 말고 머리 위치만 옮긴다.

````python
items = []
head = 0          # 맨 앞 위치 표시

for x in [10, 20, 30]:
    items.append(x)          # enqueue

x = items[head]              # dequeue
head += 1
print(x)                     # 10

x = items[head]
head += 1
print(x)                     # 20

print(head)                  # 2
print(items)                 # [10, 20, 30]  데이터는 그대로
print(head >= len(items))    # False  is_empty
````

자세한 이야기는 시간 복잡도 단원에서 다룬다.
