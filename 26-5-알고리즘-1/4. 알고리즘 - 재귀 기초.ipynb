# 4. 알고리즘 - 재귀 기초

함수가 자기 자신을 다시 부르는 것을 **재귀(recursion)** 라고 한다.
스택/큐에서 배운 **호출 스택**이 바로 재귀의 동작 원리다. 함수를 부를 때마다 스택에 쌓이고,
`return`으로 끝나면 스택에서 하나씩 빠진다.

## 재귀의 두 부분
- **기저 조건(base case):** 더 이상 재귀하지 않고 값을 바로 돌려주는 멈춤 지점. **이게 없으면 무한 재귀.**
- **재귀 호출(recursive call):** 문제를 더 작은 문제로 줄여서 자기 자신을 부르는 부분.

## 푸는 방법
- **[직접 풀기]** 칸: 빈칸/시그니처만 있다. 직접 채운다.
- **[정답]** 칸: 완성 코드. 풀어본 뒤 비교한다.
- **출력 예측** 문제는 실행 전에 결과를 먼저 종이에 적는다.

## 목록
1. 팩토리얼 (도입)
2. 거듭제곱
3. 배열 다루기 (합 / 최댓값)
4. 재귀 <-> 반복 변환
5. 이중 재귀 (피보나치 / 계단 / 타일링)
6. 유클리드 호제법 (수학형 재귀)

## 1. 팩토리얼 (도입)

`n! = n x (n-1) x ... x 1`. 재귀로 보면 `f(n) = n * f(n-1)`, 기저 조건은 `f(1) = 1`.

**(1-1) 출력 예측** - 아래 코드의 출력을 먼저 예측한 뒤 실행한다.

````python
def f(n):
    if n <= 1: # 0! = 1, 1!=1
        return 1
    return n * f(n-1)  # n! = n*(n-1)!

print(f(4))   # 예측: ?
print(f(5))   # 예측: ?
````

**정답:** `24`, `120`
```
f(4) = 4 * f(3) = 4 * 3 * f(2) = 4 * 3 * 2 * f(1) = 4*3*2*1 = 24
f(5) = 5 * 4 * 3 * 2 * 1 = 120
```

**(1-2) 빈칸 - 기저 조건** 재귀를 멈추는 조건을 채워라.

````python
# [직접 풀기]
def f(n):
    if ____:        # 기저 조건
        return 1
    return n * f(n-1)

# print(f(5))   # 120
# print(f(3))   # 6
````

````python
# [정답]
def f(n):
    if n <= 1:      # n <= 1 (또는 n == 1)
        return 1
    return n * f(n-1)

print(f(5))   # 120
print(f(3))   # 6
````

**(1-3) 디버깅 - 무한 재귀** 아래 함수는 무한 재귀에 빠진다. 한 곳을 고쳐라.

````python
# [직접 풀기] - 어디가 문제일까?
def f(n):
    if n == 1:
        return 1
    return n * f(n)
````

````python
# [정답]
def f(n):
    if n == 1:
        return 1
    return n * f(n-1)   # f(n) -> f(n-1): n을 줄여야 기저 조건에 도달한다

print(f(5))   # 120
````

**(1-4) 전체 작성** `n`의 팩토리얼을 재귀로 구현하라.

````python
# [직접 풀기]
def f(n):
    pass

# print(f(6))   # 720
# print(f(4))   # 24
# print(f(1))   # 1
````

````python
# [정답]
def f(n):
    if n <= 1:
        return 1
    return n * f(n-1)

print(f(6))   # 720
print(f(4))   # 24
print(f(1))   # 1
````

## 2. 거듭제곱

`base^exp`. 재귀 관계: `base^exp = base * base^(exp-1)`, 기저 조건은 `exp == 0`일 때 `1`.

**(2-1) 빈칸 - 재귀 호출** return 부분을 채워라.

````python
# [직접 풀기]
def f(base, exp):
    if exp == 0:
        return 1
    
    return base * f(base, exp - 1)

# print(f(2, 3))   # 8
# print(f(3, 4))   # 81
f(3, 1)
````

````python
# [정답]
def f(base, exp):
    if exp == 0:
        return 1
    return base * f(base, exp-1)

print(f(2, 3))   # 8
print(f(3, 4))   # 81
````

**(2-2) 트레이싱** 아래 함수로 `f(2, 4)`를 실행하면 `f`는 **총 몇 번** 호출될까? (실행 전에 세어보기)

````python
def f(base, exp):
    if exp == 0:
        return 1
    return base * f(base, exp-1)

print(f(2, 4))   # 값은 16. 그런데 호출은 몇 번?
````

**정답:** **5번**.
```
f(2,4) -> f(2,3) -> f(2,2) -> f(2,1) -> f(2,0)
                                        (기저 조건)
```
`exp`가 4,3,2,1,0 로 다섯 번 호출된다.

## 3. 배열 다루기 (합 / 최댓값)

배열도 "맨 뒤 원소 하나 + 나머지"로 쪼개면 재귀가 된다. `n`은 처리할 원소 개수.

**(3-1) 빈칸 - 배열 합** `arr`의 앞 `n`개 원소의 합. 빈칸을 채워라.

````python
# [직접 풀기]
def f(arr, n):
    if n == 0:
        return 0
    return arr[n-1] + f(arr, n-1)

# print(f([1, 2, 3, 4], 3))   # 10
# print(f([5, 10, 15], 3))    # 30
````

````python
# [정답]
def f(arr, n):
    if n == 0:
        return 0
    return arr[n-1] + f(arr, n-1)   # 현재 원소 + 나머지의 합

print(f([1, 2, 3, 4], 4))   # 10
print(f([5, 10, 15], 3))    # 30
````

**(3-2) 전체 작성 - 배열 최댓값** 앞 `n`개 원소의 최댓값을 재귀로 구하라.

````python
# [직접 풀기]
def f(arr, n):
    pass

# print(f([3, 1, 4, 1, 5], 5))   # 5
# print(f([10, 20, 5], 3))       # 20
````

````python
# [정답]
def f(arr, n):
    if n == 1:
        return arr[0]
    return max(arr[n-1], f(arr, n-1))   # 현재 원소 vs 나머지 중 최댓값

print(f([3, 1, 4, 1, 5], 5))   # 5
print(f([10, 20, 5], 3))       # 20
````

## 4. 재귀 <-> 반복 변환

모든 재귀는 반복(for/while)으로 바꿀 수 있다. 팩토리얼을 for문으로 바꿔보자.

**(4-1) 블록 채우기** for문 내부를 채워 팩토리얼을 완성하라.

````python
# [직접 풀기]
def f(n):
    result = 1
    for i in range(1, n+1):
        result *= i
    return result

# print(f(5))   # 120
# print(f(4))   # 24
````

````python
# [정답]
def f(n):
    result = 1
    for i in range(1, n+1):
        result = result * i   # 1*2*3*...*n
    return result

print(f(5))   # 120
print(f(4))   # 24
````

## 5. 이중 재귀 (피보나치 / 계단 / 타일링)

한 번에 재귀를 **두 번** 부르는 형태: `f(n) = f(n-1) + f(n-2)`.
피보나치, 계단 오르기, 2xn 타일링이 모두 같은 점화식이다(시작값만 다름).

**(5-1) 출력 예측** 피보나치. 출력을 먼저 예측하라.

````python
def f(n):
    if n <= 2:
        return 1
    return f(n-1) + f(n-2)

# 1 1 2 3 5 8 13
print(f(6))   # 예측: ?
print(f(7))   # 예측: ?
````

**정답:** `8`, `13`
```
피보나치: 0, 1, 1, 2, 3, 5, 8, 13, ...
f(6)=8, f(7)=13
```

**(5-2) 빈칸 - 기저 조건** 두 기저 조건을 채워라.

````python
# [직접 풀기]
def f(n):
    if n == 0:
        return ____
    if n == 1:
        return ____
    return f(n-1) + f(n-2)

# print(f(5))   # 5
# print(f(8))   # 21
````

````python
# [정답]
def f(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return f(n-1) + f(n-2)

print(f(5))   # 5
print(f(8))   # 21
````

**(5-3) 응용 - 계단 오르기 / 타일링** 한 번에 1칸 또는 2칸 오를 때 `n`칸 계단을 오르는 방법의 수.
(`n=1`이면 1가지, `n=2`이면 2가지) 재귀로 작성하라. 2xn 타일링도 똑같은 코드다.

````python
# [직접 풀기]
def f(n):
    pass

# print(f(5))   # 8
# print(f(6))   # 13
````

````python
# [정답]
def f(n):
    if n == 1:
        return 1
    if n == 2:
        return 2
    return f(n-1) + f(n-2)

print(f(5))   # 8
print(f(6))   # 13
````

**(5-4) 트레이싱 - 중복 호출** 위 피보나치로 `f(5)`를 실행하면 `f`가 총 몇 번 호출될까?
직접 세보고, 왜 이렇게 많은지 생각해보라.

````python
# 호출 횟수를 세어보는 코드
# f(5) => f(4) + f(3)  -> 4+2+2
# f(4) => f(3) + f(2)  -> 4
# f(3) => f(2) + f(1)  -> 2
count = 0
def f(n):
    global count
    count += 1
    if n <= 1:
        return n
    return f(n-1) + f(n-2)

f(5)
print(count)   # ?
````

**정답:** **15번**.
```
              f(5)
           /        \
        f(4)          f(3)
       /    \        /    \
    f(3)   f(2)   f(2)   f(1)
    ...
```
`f(3)`을 2번, `f(2)`를 3번 중복 계산한다. 이중 재귀는 같은 값을 여러 번 다시 계산해서
`n`이 커지면 폭발적으로 느려진다(나중에 배울 시간복잡도 주제로 이어짐).

## 6. 유클리드 호제법 (수학형 재귀)

두 수의 최대공약수(gcd)를 구하는 고전 알고리즘.
핵심 성질: **`gcd(a, b) = gcd(b, a % b)`**, 그리고 `b == 0`이면 `gcd(a, 0) = a`.

**(6-1) 빈칸 - 재귀 gcd** 기저 조건과 재귀 호출을 채워라.

````python
# [직접 풀기]
def f(a, b):
    if ____:
        return a
    return f(____, ____)

# print(f(24, 36))   # 12
# print(f(15, 25))   # 5
````

````python
# [정답]
def f(a, b):
    if b == 0:
        return a
    return f(b, a % b)

print(f(24, 36))   # 12
print(f(15, 25))   # 5
````

**(6-2) 트레이싱** `f(48, 18)`을 계산할 때 `(a, b)`가 어떻게 바뀌는지 나열하고, 재귀가 몇 번 일어나는지 세어라.

````python
def f(a, b):
    print("f(", a, ",", b, ")")   # 호출될 때마다 출력
    if b == 0:
        return a
    return f(b, a % b)

print("결과:", f(48, 18))
````

**정답:**
```
f(48, 18) -> f(18, 12) -> f(12, 6) -> f(6, 0) -> 결과 6
```
`b`가 18, 12, 6, 0 으로 바뀌며 재귀는 3번 일어난다(`b == 0`에서 멈춤).
나머지 연산으로 값이 빠르게 줄어서 매우 효율적이다.

## 정리

- 재귀 = **기저 조건 + 재귀 호출**. 기저 조건이 없거나 값을 줄이지 않으면 무한 재귀.
- 함수 호출은 **스택**에 쌓인다(스택/큐 단원과 연결).
- 선형 재귀(팩토리얼, 배열): 호출 1번. 이중 재귀(피보나치): 호출 2번 -> 중복 계산 많음.
- 모든 재귀는 반복으로 바꿀 수 있다.
