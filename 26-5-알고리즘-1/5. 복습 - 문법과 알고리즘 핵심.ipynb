# 5. 복습 - 문법과 알고리즘 핵심

지금까지 배운 것을 **문법 -> 알고리즘** 순서로 다시 짚는다.
알고리즘이 안 풀리는 이유는 대부분 알고리즘을 몰라서가 아니라 **문법이 손에 안 붙어서**다.
그래서 1부에서 문법을 손으로 다시 굴리고, 2부에서 스택/큐/재귀를 다시 본다.

## 푸는 방법
- **출력 예측**: 실행 버튼을 누르기 전에 **먼저 답을 적거나 말한다.** 그다음 실행해서 맞는지 본다.
- **[직접 풀기]**: 빈칸 `____` 또는 `pass`를 직접 채운다.
- **[정답]**: 다 풀어본 뒤에 비교한다. 먼저 보면 아무것도 안 남는다.

## 목차
**1부. 문법**

8. 큐 (리스트로 직접)

## 2. 리스트 인덱싱과 슬라이싱

리스트는 **0번부터** 센다. 마지막은 `-1`.
슬라이싱 `arr[a:b]`는 **a부터 b 직전까지** (b는 포함 안 함).

**(2-1) 출력 예측**

````python
arr = [10, 20, 30, 40, 50]

print(arr[0])
print(arr[4])
print(arr[-1])
print(arr[-2])
print(len(arr))
# print(arr[5])   # 주석 풀면? -> 예측
````

**정답:** `10` / `50` / `50` / `40` / `5`

`arr[5]`는 **IndexError**. 길이가 5면 유효한 인덱스는 0~4다.
`arr[len(arr)]`은 항상 에러, `arr[len(arr)-1]`이 마지막.

**(2-2) 출력 예측 - 슬라이싱**

````python
arr = [10, 20, 30, 40, 50]

print(arr[1:3])
print(arr[:2])
print(arr[3:])
print(arr[:])
print(arr[::-1])
````

**정답:** `[20, 30]` / `[10, 20]` / `[40, 50]` / `[10, 20, 30, 40, 50]` / `[50, 40, 30, 20, 10]`

- `arr[1:3]`은 1번, 2번만. **3번은 안 들어간다.**
- 앞을 생략하면 처음부터, 뒤를 생략하면 끝까지.
- `[::-1]`은 뒤집기.

**(2-3) 출력 예측 - 리스트를 바꾸는 메서드** 각 줄 실행 후 `arr`의 상태를 추적하라.

````python
arr = [1, 2, 3]

arr.append(4)
print(arr)

arr.insert(0, 0)
print(arr)

x = arr.pop()
print(x, arr)

y = arr.pop(0)
print(y, arr)
````

**정답:**
```
[1, 2, 3, 4]
[0, 1, 2, 3, 4]
4 [0, 1, 2, 3]
0 [1, 2, 3]
```
- `append(x)` 맨 뒤에 추가
- `insert(0, x)` 맨 앞에 삽입
- `pop()` 맨 뒤를 **꺼내면서 지움** (지운 값을 돌려준다)
- `pop(0)` 맨 앞을 꺼내면서 지움

이 네 개가 곧 스택과 큐다.

**(2-4) 함정 - 반환값이 없는 메서드** 출력을 예측하라.

````python
arr = [3, 1, 2]
r = arr.sort()
print(arr)
print(r)

b = sorted([3, 1, 2])
print(b)
````

**정답:** `[1, 2, 3]` / `None` / `[1, 2, 3]`

`arr.sort()`는 **arr 자체를 정렬하고 아무것도 돌려주지 않는다**(`None`).
`sorted(arr)`는 원본을 두고 **정렬된 새 리스트를 돌려준다**.

`arr = arr.sort()` 라고 쓰면 arr이 `None`이 되어버린다. 흔한 실수다.
`append`, `insert`도 마찬가지로 `None`을 돌려준다.

## 3. 반복문 (for / range / while)

`range(a, b)`는 **a부터 b 직전까지**. 슬라이싱과 같은 규칙이다.

**(3-1) 출력 예측** 각각 몇 번 돌고 무엇이 나오는가?

````python
for i in range(3):
    print("A", i)

for i in range(1, 4):
    print("B", i)

for i in range(0, 10, 3):
    print("C", i)

for i in range(3, 0, -1):
    print("D", i)
````

**정답:**
```
A 0, A 1, A 2
B 1, B 2, B 3
C 0, C 3, C 6, C 9
D 3, D 2, D 1
```
- `range(n)` -> `0 ~ n-1` (**n번** 반복)
- `range(a, b)` -> `a ~ b-1`
- 세 번째 인자는 간격. 음수면 거꾸로.

**(3-2) 출력 예측 - 리스트 순회 두 방식**

````python
arr = [10, 20, 30]

for x in arr:
    print("값:", x)

for i in range(len(arr)):
    print("인덱스", i, "값", arr[i])
````

**정답:** 위는 값 10,20,30. 아래는 인덱스와 값을 같이.

**인덱스가 필요 없으면 `for x in arr`**를 쓴다.
인덱스가 필요할 때(앞뒤 원소 비교, 위치 저장 등)만 `range(len(arr))`.

## 4. 조건문과 논리 연산

**(4-1) 출력 예측 - if / elif / else** 각 입력에서 무엇이 출력되나?

````python
def grade(n):
    if n >= 90:
        return "A"
    elif n >= 80:
        return "B"
    elif n >= 70:
        return "C"
    else:
        return "F"

print(grade(95), grade(85), grade(70), grade(60))
````

**정답:** `A B C F`

`elif`는 **앞 조건이 거짓일 때만** 검사한다. 그래서 `n >= 80`은 사실상 `80 <= n < 90`이다.
만약 전부 `if`로 쓰면 95점이 A, B, C에 모두 걸린다(마지막 것이 반환됨).

**(4-2) 출력 예측 - and / or / not**

````python
a, b = 5, 0

print(a > 0 and b > 0)
print(a > 0 or b > 0)
print(not (a > 0))
print(a > 0 and a < 10)
print(0 < a < 10)
````

**정답:** `False` / `True` / `False` / `True` / `True`

- `and`는 **둘 다** 참이어야 참
- `or`는 **하나라도** 참이면 참
- 파이썬은 `0 < a < 10` 처럼 이어쓰기가 된다 (다른 언어에선 안 된다)

**(4-3) 함정 - 단락 평가** 3번 노트북에서 본 것. 빈 리스트일 때 (A)와 (B)는?

````python
s = []

print(len(s) == 0 or s.pop() != '(')     # (A)
# print(s.pop() != '(' or len(s) == 0)   # (B) 주석 풀면?
````

**정답:** (A) `True`. (B)는 **IndexError**.

`or`는 왼쪽이 참이면 오른쪽을 **실행하지 않는다**(단락 평가).
그래서 **위험한 연산 앞에 안전 검사를 먼저** 둔다. 스택이 비었는지 먼저 보고 pop.
`and`도 같다. 왼쪽이 거짓이면 오른쪽을 안 본다: `if len(s) > 0 and s[-1] == '('`.

**(4-4) 출력 예측 - 빈 것은 거짓** 파이썬에서 무엇이 거짓 취급되나?

````python
print(bool([]), bool([0]))
print(bool(""), bool("0"))
print(bool(0), bool(1), bool(-1))

s = []
if not s:
    print("비었다")
````

**정답:** `False True` / `False True` / `False True True` / `비었다`

**빈 리스트, 빈 문자열, 숫자 0**이 거짓. 나머지는 참이다.
`"0"`은 **비어있지 않은 문자열**이라 참이다. 주의.

그래서 `if len(s) == 0:` 대신 `if not s:` 라고 쓸 수 있다. (둘 다 맞다)

## 5. 함수 (return의 의미)

가장 자주 틀리는 부분: **`print`는 화면에 찍는 것, `return`은 값을 돌려주는 것.**

**(5-1) 출력 예측** 두 함수의 차이를 보라.

````python
def f1(n):
    print(n * 2)

def f2(n):
    return n * 2

a = f1(5)
b = f2(5)

print("a =", a)
print("b =", b)
print(f2(5) + 1)
# print(f1(5) + 1)   # 주석 풀면?
````

**정답:**
```
10          <- f1이 직접 찍은 것
a = None    <- f1은 돌려주는 값이 없다
b = 10
11
```
`f1(5) + 1`은 **TypeError** (`None + 1`).

**`return`이 없는 함수는 `None`을 돌려준다.** 계산 결과를 다른 곳에 쓰려면 반드시 `return`.

**(5-2) 출력 예측 - return은 즉시 종료**

````python
def f(n):
    if n > 0:
        return "양수"
    print("여기 지나감")
    return "음수 또는 0"

print(f(5))
print(f(-5))
````

**정답:**
```
양수
여기 지나감
음수 또는 0
```
`f(5)`에서는 `return`을 만나는 순간 함수가 끝나서 `print("여기 지나감")`이 실행되지 않는다.
`return`은 값을 돌려주면서 **함수를 즉시 빠져나온다.**

## 7. 스택 (리스트로 직접)

**LIFO** - 마지막에 넣은 것이 먼저 나온다.
- 넣기 `s.append(x)`
- 꺼내기 `s.pop()`  <- 인자 없음

```
    -> pop:4 -> [1,2] -> pop:2 -> [1]
```

**(7-2) 빈칸 - 괄호 검사** 가장 대표적인 스택 문제. 빈칸을 채워라.

````python
# [직접 풀기]
expr = "(()())"
s = []
ok = True

for ch in expr:
    if ch == '(':
        s.____(ch)               # 여는 괄호면?
    elif ch == ')':
        if len(s) == 0:          # 닫는데 열린 게 없다
            ok = False
            break
        s.____()                 # 짝을 맞춰 하나 제거
if ____:                         # 마지막에 스택에 남은 게 있으면?
    ok = False

print(ok)   # True
````

````python
# [정답]
expr = "(()())"
s = []
ok = True

for ch in expr:
    if ch == '(':
        s.append(ch)
    elif ch == ')':
        if len(s) == 0:
            ok = False
            break
        s.pop()
if len(s) > 0:        # 열고 안 닫은 게 남아있다
    ok = False

print(ok)   # True

# 확인
for t in ["(())", "(()", "())", "", "()()"]:
    s = []
    ok = True
    for ch in t:
        if ch == '(':
            s.append(ch)
        elif ch == ')':
            if len(s) == 0:
                ok = False
                break
            s.pop()
    if len(s) > 0:
        ok = False
    print(repr(t), ok)
````

**핵심 두 가지 - 이걸 빠뜨려서 틀린다:**
1. 닫는 괄호가 왔는데 **스택이 비었으면** 틀린 것 -> `())`
2. 다 돌고 나서 **스택에 남아있으면** 틀린 것 -> `(()`

**(7-3) 출력 예측 - 후위식 계산** `pop` 두 번의 순서에 주의.

````python
expr = "5 1 2 + 4 * + 3 -"
s = []

for t in expr.split():
    if t in ('+', '-', '*', '/'):
        b = s.pop()      # 오른쪽 피연산자
        a = s.pop()      # 왼쪽 피연산자
        if t == '+': s.append(a + b)
        elif t == '-': s.append(a - b)
        elif t == '*': s.append(a * b)
        else: s.append(a / b)
    else:
        s.append(int(t))

print(s.pop())
````

**정답:** `14`
```
5 1 2 + 4 * + 3 -
= 5 + ((1+2) * 4) - 3
= 5 + 12 - 3 = 14
```
`b`를 먼저 꺼내는 이유: 나중에 넣은 것이 먼저 나오므로 **먼저 꺼낸 것이 오른쪽**이다.
순서를 바꾸면 `-`와 `/`에서 답이 틀린다.

## 8. 큐 (리스트로 직접)

**FIFO** - 먼저 넣은 것이 먼저 나온다.
- 넣기 `q.append(x)`
- 꺼내기 `q.pop(0)`  <- **0을 넣는다**

**(8-1) 출력 예측**

````python
q = []
q.append(1)
q.append(2)
q.append(3)
print(q.pop(0))
q.append(4)
print(q.pop(0))
print(q)
````

**정답:** `1` / `2` / `[3, 4]`

스택이었다면 `3`, `4`가 나왔을 것이다. **`pop()`과 `pop(0)` 한 글자 차이**로 완전히 달라진다.

**(8-2) 빈칸 - 카드 버리기(요세푸스)** 맨 위 카드는 버리고, 다음 카드는 맨 아래로 보낸다. 마지막에 남는 카드는?

````python
# [직접 풀기]
q = [1, 2, 3, 4, 5, 6]

while len(q) > 1:
    q.____(0)              # 맨 위 카드를 버린다
    q.append(q.____(0))    # 다음 카드를 맨 아래로

print(q[0])   # 4
````

````python
# [정답]
q = [1, 2, 3, 4, 5, 6]

while len(q) > 1:
    q.pop(0)               # 버림
    q.append(q.pop(0))     # 꺼내서 뒤에 붙임 = 맨 아래로 이동

print(q[0])   # 4

# 과정: [1..6] -> 1버림, 2뒤로 -> [3,4,5,6,2] -> 3버림, 4뒤로 -> [5,6,2,4]
#      -> 5버림, 6뒤로 -> [2,4,6] -> 2버림, 4뒤로 -> [6,4] -> 6버림, 4뒤로 -> [4]
````

**(8-3) 출력 예측 - 진법 변환** 큐에 든 자릿수를 값으로 만든다.

````python
q = [1, 0, 2, 1]   # 3진수 (앞이 높은 자리)
b = 3
t = 0
while len(q) > 0:
    t = b * t + q.pop(0)
print(t)
````

**정답:** `34`. `((1*3+0)*3+2)*3+1 = 34`

한 자리 꺼낼 때마다 **지금까지 값에 진법을 곱하고**(자리 올림) 새 자릿수를 더한다.

## 9. 재귀

**재귀 = 기저 조건 + 자기 호출.** 둘 중 하나라도 잘못되면 무한 재귀(RecursionError).
- **기저 조건**: 더 안 부르고 값을 바로 돌려주는 지점
- **자기 호출**: 반드시 **인자를 줄여서** 기저 조건 쪽으로 가야 한다

**(9-1) 출력 예측**

````python
def f(n):
    if n <= 1:
        return 1
    return n * f(n-1)

print(f(4))
print(f(1))
print(f(0))
````

**정답:** `24` / `1` / `1`

`f(0)`도 `n <= 1`에 걸려서 1이다. 기저 조건을 `n == 1`로 썼다면 `f(0)`은 **무한 재귀**가 된다
(0 -> -1 -> -2 -> ... 영원히 1에 도달하지 못한다).

**(9-2) 디버깅** 아래 두 함수는 각각 어디가 잘못됐나? 고쳐라.

````python
# [직접 풀기] - 두 개 다 고칠 것

def f1(n):          # 팩토리얼
    if n == 1:
        return 1
    return n * f1(n)

def f2(n):          # 1부터 n까지의 합
    return n + f2(n-1)
````

````python
# [정답]
def f1(n):
    if n <= 1:
        return 1
    return n * f1(n-1)   # n -> n-1 : 인자를 줄여야 기저 조건에 도달한다

def f2(n):
    if n == 0:           # 기저 조건 자체가 없었다
        return 0
    return n + f2(n-1)

print(f1(5))   # 120
print(f2(5))   # 15
````

**(9-4) 트레이싱 - 피보나치의 중복** `f(5)`를 부르면 `f`는 몇 번 호출되나?

````python
count = 0
def f(n):
    global count
    count += 1
    if n <= 1:
        return n
    return f(n-1) + f(n-2)

print(f(5), count)
````

**정답:** `5 15` - 값은 5인데 호출은 **15번**.

`f(3)`을 2번, `f(2)`를 3번 중복 계산한다. 이중 재귀는 `n`이 조금만 커져도 폭발한다.
(`f(30)`이면 약 270만 번)

````python
# [정답]
def show_reverse(arr, n):
    if n == 0:          # 남은 원소가 없으면 끝
        return
    print(arr[n-1])     # 마지막 원소 먼저 출력
    show_reverse(arr, n-1)   # 나머지를 재귀

show_reverse([1,2,3,4], 4)

# 순서를 바꾸면 정방향이 된다 (출력을 재귀 뒤로 미룬다)
def show_forward(arr, n):
    if n == 0:
        return
    show_forward(arr, n-1)
    print(arr[n-1])

print("---")
show_forward([1,2,3,4], 4)
````

`print`를 재귀 **앞**에 두느냐 **뒤**에 두느냐로 순서가 뒤집힌다.
재귀 뒤에 두면 "가장 깊이 들어간 것부터 돌아오면서" 실행되기 때문이다(호출 스택).
